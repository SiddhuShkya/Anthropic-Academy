## Exercise 1: Chat Bot 

> Video Link: [Chat Bot Exercise](https://anthropic.skilljar.com/claude-with-the-anthropic-api/287727)

---

Make a chatbot using the 3 helper functions we just put together.

1. Prompt the user to enter some input using the built-in "input" function

2. Add it to a list of messages.

3. Call the API.

4. Add generated text to the listg of messages.

5. Print the generated text

6. Repeat from #1

```python
# HELPER FUNCTIONS

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
    )
    return message.content[0].text
```

```python
# CHAT BOT LOGIC

## Make an initial set of messages
messages = []

## Make a while true loop to run the Chat Bot
while True:
    ## Get user input
    user_input = input("> ")
    print(">", user_input)
    ## Add user input to the list of messages
    add_user_message(messages, user_input)
    ## Call claude using the chat function
    answer = chat(messages)
    add_assistant_message(messages, answer)
    # Print the generated text
    print("---")
    print(answer)
    print('---')
```

## Exercise 2: System Prompt

> Video Link: [System Prompt Exercise](https://anthropic.skilljar.com/claude-with-the-anthropic-api/287724)

---

Try to reduce the amount of code that's being generated when we prompt:

"Write a python function that checks a string for duplicate characters."

Without system prompt, the chatbot will generate a little bit of code, but there's also going to be a lot of explanation, and a lot of comments as well.

Create a system prompt so that the AI generated  response will reduce the amount of code being generated, bringing it down to as concies and implementation of this function as possible.

The system prompt should assign a role to Claude and encourage it to respond as concisely as possible.

```python
# HELPER FUNCTIONS

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    if system:
        params['system'] = system
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

add_user_message(
    messages,
    "Write a python function that checks a string for duplicate characters."
)

answer = chat(
    messages, 
    system="You are a Python Engineer who writes very concise code."
)
```
