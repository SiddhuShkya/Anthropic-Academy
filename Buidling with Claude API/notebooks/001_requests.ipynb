```python
# CREATE AN API CLIENT
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-5"
```

```python
# MAKE A REQUEST
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is vibe coding? Explain in 2 sentences."
        }
    ]
)
```
